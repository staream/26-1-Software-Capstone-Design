```python
import os
import shutil
from pathlib import Path

src_img_dir= Path("/content/drive/MyDrive/TDA/dataset/images/train")
original_source_label= Path("/content/drive/MyDrive/TDA/Final_Pipeline/labels")
work_dir = Path("/content/drive/MyDrive/TDA/Final_Pipeline")
dst_img_dir = work_dir /"img_ori"
# dst_src_label= work_dir/"label"

# Experiments
signal_dir = work_dir/"signal"
sobel_dir = work_dir/"sobel"
PH_dir = work_dir/"PH"
PH_Sobel_dir = work_dir/"PH_sobel"

# Results
detect_result_dir = work_dir /"detection_results"
migration_result_dir = work_dir / "migration_results"


for d in [dst_img_dir, signal_dir, sobel_dir, PH_dir, PH_Sobel_dir,
          detect_result_dir, migration_result_dir]:
    d.mkdir(parents=True, exist_ok=True)

print("폴더 준비 완료")
print("원본 이미지 폴더:", src_img_dir)
print("작업 이미지 폴더:", dst_img_dir)
```

```python
image_exts = [".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff"]

# label 파일들의 이름만 추출
label_stems = {label_path.stem for label_path in original_source_label.iterdir()
               if label_path.is_file()}
print("라벨 개수:", len(label_stems))

copied_count = 0
missing_count = 0
missing_names = []

for stem in label_stems:
    found = False

    for ext in image_exts:
        src_img_path = src_img_dir / f"{stem}{ext}"

        if src_img_path.exists():
            dst_img_path = dst_img_dir / src_img_path.name
            shutil.copy2(src_img_path, dst_img_path)

            copied_count += 1
            found = True
            break

    if not found:
        missing_count += 1
        missing_names.append(stem)

print("복사 완료")
print("복사된 이미지 수:", copied_count)
print("이미지를 찾지 못한 라벨 수:", missing_count)

if missing_names:
    print("찾지 못한 파일 예시:", missing_names[:10])
```
